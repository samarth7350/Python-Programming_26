```python
# Program to find largest between three numbers.

a = int(input("Enter number a :"))
b = int(input("Enter number b :"))
c = int(input("Enter number c :"))

print("Numbers are :", a, ",", b, ",", c)

if a>b and a>c:
    print("a is the largest.")
elif b>a and b>c:
    print("b is the largest.")
elif c>a and c>b:
    print("c is the largest.")
elif a==b and a>c:
    print("a and b are the largest and equal.")
elif a==c and a>b:
    print("a and c are the largest and equal.")
elif b==c and b>a:
    print("b and c are the largest and equal.")
else:
    print("a, b, and c are equal.")
```
