```python
# Program to perform list operations on students list.


# Original list
studentlist = ["Aditya", "Bhargav", "Chinmay", "Darshan", "Eshwar", "Farhan", "Gaurav", "Harsh"]
print(studentlist)


# Add operations
studentlist.append("Ishan")
print(studentlist)

studentlist.insert(1, "Atharva")
print(studentlist)


# Delete operations
studentlist.remove("Aditya")
print(studentlist)

studentlist.pop(-1)
print(studentlist)

del studentlist[-1]
print(studentlist)


# Update operations
studentlist[0] = "Aditya"
print(studentlist)

studentlist[1:3] = ["Bhuvan", "Chaitanya"]
print("List :", studentlist)
```

```python
# Program to perform tuple operations on students list.


# Original tuple
studenttuple = ("Aditya", "Bhargav", "Chinmay", "Darshan", "Eshwar", "Farhan", "Gaurav", "Harsh")
print(studenttuple)


# Convert tuple into list
studentlist = list(studenttuple)


# Add operations
studentlist.append("Ishan")
print(studentlist)

studentlist.insert(1, "Atharva")
print(studentlist)


# Delete operations
studentlist.remove("Aditya")
print(studentlist)

studentlist.pop(-1)
print(studentlist)

del studentlist[-1]
print(studentlist)


# Update operations
studentlist[0] = "Aditya"
print(studentlist)

studentlist[1:3] = ["Bhuvan", "Chaitanya"]
print(studentlist)


# Convert list into tuple
studenttuple = tuple(studentlist)

print("Tuple :", studenttuple)
```

```python
# Program to perform dictionary operations on student dictionary.


# Original dictionary
studentdictionary = {
    "Name" : "Aditya",
    "PRN no" : 1272261234,
    "Age" : 18,
}
print(studentdictionary)


# Add operations
studentdictionary["Phone no"] = 9800000000
print(studentdictionary)

studentdictionary.update({"Email" : "aditya1234@gmail.com"})
print(studentdictionary)


# Delete operations
studentdictionary.pop("Age")
print(studentdictionary)

studentdictionary.popitem()
print(studentdictionary)

del studentdictionary["Phone no"]
print(studentdictionary)


# Update operations
studentdictionary["Name"] = "Bhargav"
print(studentdictionary)

studentdictionary.update({"PRN no" : 1272264321})
print(studentdictionary)
```
