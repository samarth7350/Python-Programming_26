```python
#3
my_list = [10, 20, 30, 40, 50]

my_tuple = tuple(my_list)

print("Tuple:", my_tuple)
```

```python
#4
my_tuple = (10, 20, 30, 40, 50)

item = int(input("Enter the item: "))

if item in my_tuple:
    print("Index of the item:", my_tuple.index(item))
else:
    print("Item not found in the tuple")
```

```python
#5
numbers = [(10, 20, 40), (40, 50, 60), (70, 80, 90)]

new_list = []

for tup in numbers:
    new_tup = tup[:-1] + (100,)
    new_list.append(new_tup)

print("Original List:", numbers)
print("Updated List:", new_list)
```

```python
#6
numbers = (10, 20, 5, 40, 30, 15)

print("Maximum element:", max(numbers))
print("Minimum element:", min(numbers))
```
